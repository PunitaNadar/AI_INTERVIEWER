```python
from google.colab import userdata
from groq import Groq

# Paste your Groq API key here
groq_key = userdata.get('groq_key')


client = Groq(
    api_key=groq_key
)

SYSTEM_PROMPT = """
You are ChefMate, a friendly and practical personal cooking assistant.

Help users choose recipes, use available ingredients, substitute missing
ingredients, plan meals, and troubleshoot cooking problems.

Always consider the user's available ingredients, cooking time, skill level,
equipment, servings, and dietary preferences when mentioned.

Give clear, step-by-step instructions with quantities and cooking times.
Prefer simple, affordable and realistic recipes.

If something goes wrong while cooking, explain the likely cause and how to fix it.
Ask questions only when necessary.

Be encouraging, conversational, and safety-conscious.
Your goal is to help the user successfully cook delicious food with what they have.
"""

messages = [
    {
        "role": "system",
        "content": SYSTEM_PROMPT
    }
]

print("🍳 ChefMate AI")
print("Your personal cooking assistant")
print("Type 'exit' to quit.\n")

while True:

    user_input = input("You: ")

    if user_input.lower().strip() in ["exit", "quit"]:
        print("\nChefMate: Happy cooking! 👨‍🍳")
        break

    messages.append({
        "role": "user",
        "content": user_input
    })

    try:

        completion = client.chat.completions.create(
            model="openai/gpt-oss-120b",
            messages=messages,
            temperature=1,
            max_completion_tokens=2048,
            top_p=1,
            reasoning_effort="medium",
            stream=True,
            stop=None
        )

        print("\nChefMate: ", end="")

        assistant_response = ""

        for chunk in completion:
            text = chunk.choices[0].delta.content or ""
            print(text, end="", flush=True)
            assistant_response += text

        print("\n")

        # Save response so ChefMate remembers the conversation
        messages.append({
            "role": "assistant",
            "content": assistant_response
        })

    except Exception as e:
        print("\nError:", e)
```
